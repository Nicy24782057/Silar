## 3. Periksa dependency inference

Profil mengikuti loader yang sudah lulus di Colab. Bila diminta restart session, restart lalu jalankan notebook dari awal.

```python
PROFILE = json.loads((ROOT/'config/inference_environment.json').read_text())
expected = PROFILE['pinned_packages']
observed = {}
for name in PROFILE['guard_packages']:
    try: observed[name] = importlib.metadata.version(name)
    except importlib.metadata.PackageNotFoundError: observed[name] = None
different = [n for n in observed if not observed[n] or observed[n].split('+',1)[0] != expected[n]]
loaded = [n for n in different if n in sys.modules]
if different:
    result = subprocess.run([sys.executable,'-m','pip','install','--disable-pip-version-check','-r',str(ROOT/'config/requirements-inference.txt')],capture_output=True,text=True)
    (OUT/'dependency_install.txt').write_text(result.stdout+result.stderr)
    print((result.stdout+result.stderr)[-5000:])
    if result.returncode: raise RuntimeError('Instalasi dependency gagal; model belum dijalankan')
    importlib.invalidate_caches()
    if loaded: raise RuntimeError('Restart session dan jalankan dari awal; modul lama sudah dimuat: '+str(loaded))
else: (OUT/'dependency_install.txt').write_text('Profil sudah sesuai; tidak menginstal ulang.\n')
sys.path.insert(0,str(ROOT/'code/active_nara_adapter'))
from slr09_candidate_runtime import check_dependencies
write(OUT/'dependency_versions.json',check_dependencies(PROFILE))
print('Dependency sesuai profil.')
```

## 5. Bekukan acuan, jalankan model, hitung mutu, ekspor bukti

Status SUCCEEDED berarti eksekusi selesai. Periksa `metric_targets_met` untuk kelulusan target; persetujuan akhir kelompok dan status Done tidak diubah otomatis. Colab tetap REPRESENTATIVE. ZIP hasil tidak menyertakan bobot.

```python
from types import SimpleNamespace
import shutil, traceback, zipfile
RESULT = OUT / 'evaluation'
failure = None
try:
    status = run(SimpleNamespace(root=ROOT,review=REVIEW,output=RESULT,sena_zip=SENA_ZIP,nara_zip=NARA_ZIP,
        device='cpu',environment_label='GOOGLE_COLAB_CPU',validate_only=False))
    report = json.loads((RESULT/'evaluation.json').read_text())
    print('Status eksekusi:',status['status'])
    print('Target mutu terpenuhi:',report['metric_targets_met'])
    print('Hasil per kriteria:',json.dumps(report['checks'],indent=2))
    print('Keputusan teknis:',report['decision'],'; persetujuan akhir kelompok tetap dicatat terpisah.')
except Exception as error:
    failure = error
    print('EVALUASI GAGAL:',error)
    (OUT/'notebook_error.txt').write_text(traceback.format_exc())
finally:
    shutil.copy2(REVIEW,OUT/REVIEW.name)
    shutil.copy2(REVIEW_XLSX,OUT/REVIEW_XLSX.name)
    shutil.copy2(REVIEW_XLSX,OUT/REVIEW_XLSX.name)
    shutil.copytree(ROOT,OUT/'source_snapshot',dirs_exist_ok=True)
    archive = WORK / ('SLR09_Hasil_Evaluasi_Penerimaan_v04_'+RUN_ID+'.zip')
    with zipfile.ZipFile(archive,'w',compression=zipfile.ZIP_DEFLATED) as z:
        for p in sorted(OUT.rglob('*')):
            if p.is_file() and 'runtime' not in p.relative_to(OUT).parts:
                z.write(p,str(p.relative_to(OUT)))
    try:
        shutil.copy2(archive,EXPORT_DIR/archive.name)
        print('Bukti tersimpan di Drive:',archive.name)
    except OSError as e: print('Salin ke Drive gagal; unduh dari panel Files:',e)
    try: files.download(str(archive))
    except Exception as e: print('Unduh manual dari panel Files:',archive,';',e)
if failure is not None: raise failure
```
