# SENA fokus NON_EVENT: eksperimen kedua CPU

Warm start dari checkpoint pengembangan terbaik eksperimen pertama. Tetap memakai split development yang sama, tanpa pilot 50 untuk fit/seleksi. Empat layer encoder akhir dilatih, LR encoder 2e-5, LR head 5e-5, pengali loss NON_EVENT 2.5. Maksimum enam epoch, early stopping patience dua, berhenti jika semua target development tercapai. Ini keputusan pengembangan setelah eksperimen pertama; bukan eksperimen yang belum pernah melihat validasi. Target tidak diturunkan.

Identitas baseline, warm start, konfigurasi dan hasil disimpan. Model ini bukan kandidat Colab lama yang bobotnya tidak tersedia.

```python
from pathlib import Path
import os,sys,json,subprocess,hashlib
ROOT=Path(os.environ['SLR09_PACKAGE_ROOT'])
ASSETS=Path(os.environ['SLR09_ASSET_DIR'])
PREVIOUS=Path(os.environ['SLR09_TRAIN_OUTPUT'])
OUTPUT=PREVIOUS.with_name('training_run_20261019_cpu_focus')
print('Source SHA:',hashlib.sha256((ROOT/'code/quality_fix/train_sena_focus_cpu.py').read_bytes()).hexdigest())
print('50 contoh lama tidak digunakan fit atau pemilihan checkpoint.')
```

```python
command=[sys.executable,str(ROOT/'code/quality_fix/train_sena_focus_cpu.py'),
 '--root',str(ROOT),'--archive',str(ASSETS/'SILAR_SENA_v1.zip'),
 '--output',str(OUTPUT),'--warm-start',str(PREVIOUS/'best_model')]
p=subprocess.Popen(command,stdout=subprocess.PIPE,stderr=subprocess.STDOUT,text=True,bufsize=1)
for line in p.stdout: print(line,end='',flush=True)
p.wait()
if p.returncode: raise RuntimeError('Eksperimen gagal; jangan klaim lulus.')
```

```python
state=json.loads((OUTPUT/'run_state.json').read_text())
metric=json.loads((OUTPUT/'best_development_metrics.json').read_text())
print(json.dumps(state,indent=2))
print(json.dumps(metric,indent=2))
assert state['status']=='SUCCEEDED' and state['training_executed']
print('Hasil development; penerimaan mutu akhir terpisah.')
```
